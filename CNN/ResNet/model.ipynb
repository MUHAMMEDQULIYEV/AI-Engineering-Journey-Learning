```python
from torch.utils.data import DataLoader
from torchvision import transforms,datasets
import torch.nn as nn
import torch
```

```python
device="cuda" if torch.cuda.is_available() else "cpu"
print(device)
```

```python
transform=transforms.Compose([transforms.ToTensor(),transforms.RandomHorizontalFlip(),transforms.RandomCrop(32,padding=4),transforms.Normalize(mean = (0.4914, 0.4822, 0.4465),
std  = (0.2470, 0.2435, 0.2616))])

test_tf=transforms.Compose([transforms.ToTensor(),transforms.Normalize(mean = (0.4914, 0.4822, 0.4465),
std  = (0.2470, 0.2435, 0.2616))])
```

```python
raw=datasets.CIFAR10(root="./data",train=True,download=True,transform=transform)
test=datasets.CIFAR10(root="./data",train=False,download=True,transform=test_tf)
```

```python
trainloader=DataLoader(raw,batch_size=128,shuffle=True)
testloader=DataLoader(test,batch_size=128,shuffle=False)
```

```python
import torch.nn.functional as F
class BasicBlock(nn.Module):
    def __init__(self,stride,in_ch,out_ch):
        super().__init__()
        self.conv1=nn.Conv2d(kernel_size=3,padding=1,in_channels=in_ch,out_channels=out_ch,stride=stride)
        self.bn1=nn.BatchNorm2d(out_ch)
        self.relu=nn.ReLU(inplace=True)
        self.conv2=nn.Conv2d(kernel_size=3,in_channels=out_ch,out_channels=out_ch,stride=1,padding=1)
        self.bn2=nn.BatchNorm2d(out_ch)
        if stride !=1 or in_ch !=out_ch:
            self.shortcut=nn.Sequential(
                nn.Conv2d(in_ch,out_ch,stride=stride,kernel_size=1,bias=False),
                nn.BatchNorm2d(out_ch),
            )
        else:
            self.shortcut=nn.Identity()



    def forward(self, x):
        identity=self.shortcut(x)
        out = self.relu(self.bn1(self.conv1(x)))
        out = self.bn2(self.conv2(out))

        out = out + identity
        return self.relu(out)
```
