```python
from torch.utils.data import DataLoader
from torchvision import transforms,datasets
import torch.nn as nn
import torch
```

```python
device="cuda" if torch.cuda.is_available() else "cpu"
print(device)
```

```python
transform=transforms.Compose([transforms.ToTensor(),transforms.RandomHorizontalFlip(),transforms.RandomCrop(32,padding=4),transforms.Normalize(mean = (0.4914, 0.4822, 0.4465),
std  = (0.2470, 0.2435, 0.2616))])

test_tf=transforms.Compose([transforms.ToTensor(),transforms.Normalize(mean = (0.4914, 0.4822, 0.4465),
std  = (0.2470, 0.2435, 0.2616))])
```

```python
raw=datasets.CIFAR10(root="./data",train=True,download=True,transform=transform)
test=datasets.CIFAR10(root="./data",train=False,download=True,transform=test_tf)
```

```python
trainloader=DataLoader(raw,batch_size=128,shuffle=True)
testloader=DataLoader(test,batch_size=128,shuffle=False)
```
