## 2. Thresholds by hand
Write `precision_recall_at(y_true, scores, threshold)` myself, **no sklearn**.
- predicted positive = score ≥ threshold
- precision = TP / flagged · recall = TP / real positives

**Predict** (from the chat): threshold 0.7 → P = ?, R = ? · 0.5 → ? · 0.3 → ?

My guess: 1,0;0.5,0.5;0.5,1

````python

def precision_recall_at(y_true,scores,threshold):
    flagged=0
    tp=0
    y_1=y_true.count(1)
    for s,y in zip(scores,y_true):
        if s >=threshold:
            if y==1:
                tp+=1
            flagged+=1
            
    recall=0 if y_1==0 else tp/y_1 
    precicion=0 if flagged==0 else tp/flagged
    return precicion,recall

for t in [0.7, 0.5, 0.3]:
    p, r = precision_recall_at(y_true, scores, t)
    print(f"threshold {t}: precision {p:.2f}, recall {r:.2f}")
precision_recall_at(y_true, scores, 0.9)
````

## 3. False positive rate (FPR)
FPR = of the real **negatives**, how many did I wrongly flag? = FP / (FP + TN)

Add FPR to my function (or write a new one).

**Predict:** FPR at threshold 0.3 = ?

My guess:0.33

````python

def false_positive_rate(y_true,scores,threshold):
    
    tn=0
    fp=0
    for s,y in zip(scores,y_true):
        if s <threshold:
            if y==0:
                tn+=1
        else:
            if y==0:
                fp+=1

    fpr=0 if (tn+fp)==0 else fp/(tn+fp)
    return fpr
print(false_positive_rate(y_true, scores, 0.5))
````

## 4. ROC curve + ROC-AUC with sklearn
ROC curve = one point (FPR, recall) for **every** threshold. AUC = area under it (0 to 1).

The idea (sklearn does the threshold sweep for me):
```python

fpr, tpr, thresholds = roc_curve(y_true, scores)   # tpr = recall
auc = roc_auc_score(y_true, scores)
```
