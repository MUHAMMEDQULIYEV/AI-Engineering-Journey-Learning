## 5. Stemming & Lemmatization (lectures 309–310)
Does `likes` → `like`? What does the vocab size become?

```python
words=["eating","eats","eaten","writing","writes","programming","programs","history","finally","finalized"]
from nltk.stem import PorterStemmer
```

```python
stem=PorterStemmer()
for word in words:
    print(word+"--->"+stem.stem(word))
```

```python
stem.stem("congratulation")
```

```python
from nltk.stem import RegexpStemmer
reg_reg=RegexpStemmer('ing$|s$|e$', min=4)
for word in words:
    print(reg_reg.stem(word))
```

## Snowball Stemming


```python
from nltk.stem import SnowballStemmer
snow_ball=SnowballStemmer("english")
for word in words:
    print(snow_ball.stem(word))
```
